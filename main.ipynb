```python
score = int(input("Enter your score: "))
if score <= 100 and score >= 0:
    if score >= 91:
        print("A")
    elif score >= 81:
        print("B")
    elif score >= 71:
        print("C")
    elif score >= 61:
        print("D")
    elif score >= 51:
        print("E")
    else:
        print("F")
```

```python
n = int(input("Enter a number: "))
sum = 0
count = 0
for i in range(1,n+1):
    if i%2==0:
        sum += i
    if i%3==0:
        count+=1
        if i%5==0:
            print("3 5")
        else:
            print("3")
    else:
        if i%5==0:
            print("5")
        else:
            print(i)
print(f"Sum of all even numbers from 1 to {n} is {sum}")
print(f"There are {count} numbers divisible by 3")
```
